```python
import os
import pysrt

def srt_to_expanded_text(srt_file_path):
    subtitles = pysrt.open(srt_file_path)
    expanded_text = ""
    
    for subtitle in subtitles:
        start_time = subtitle.start.to_time()
        end_time = subtitle.end.to_time()
        expanded_text += f"{start_time} --> {end_time}\n{subtitle.text}\n\n"
    
    return expanded_text

def process_srt_files(input_folder):
    output_folder = os.path.join(input_folder, 'output')
    os.makedirs(output_folder, exist_ok=True)

    # input_folder에서 모든 .srt 파일을 찾기
    for file_name in os.listdir(input_folder):
        if file_name.endswith('.srt'):
            srt_file_path = os.path.join(input_folder, file_name)
            txt_file_path = os.path.join(output_folder, f"{os.path.splitext(file_name)[0]}.txt")
            
            # SRT 파일을 읽어 늘어진 상태로 변환 후 TXT 파일로 저장
            expanded_text = srt_to_expanded_text(srt_file_path)
            with open(txt_file_path, 'w', encoding='utf-8') as txt_file:
                txt_file.write(expanded_text)

# 사용 예시
input_folder = 'input'  # SRT 파일들이 있는 폴더 경로
process_srt_files(input_folder)
```
