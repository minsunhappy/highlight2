```python
import requests
from bs4 import BeautifulSoup

# 웹페이지 URL (실제 URL로 변경해야 합니다)
url = "https://www.youtube.com/watch?v=Cbij3MKhdOY"

# 웹페이지 내용 가져오기
response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
elements = soup.find_all('yt-live-chat-text-message-renderer')
# Extract and print the text content of each timestamp
for timestamp in elements:
    print(timestamp.get_text())
```

```python
response = requests.get(url)

# Check if the request was successful
if response.status_code == 200:
    # Parse the HTML content with BeautifulSoup
    soup = BeautifulSoup(response.content, 'html.parser')

    # Find all elements with the specified structure
    # Here, we're looking for <yt-live-chat-text-message-renderer> tags
    elements = soup.find_all('yt-live-chat-text-message-renderer', class_='style-scope yt-live-chat-item-list-renderer')

    # Extract and print the contents of each element
    for element in elements:
        print(element)
else:
    print(f"Failed to retrieve the webpage. Status code: {response.status_code}")
```
