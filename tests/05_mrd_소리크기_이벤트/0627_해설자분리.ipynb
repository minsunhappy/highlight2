```python
from pydub import AudioSegment
import os

def split_audio(audio_file, chunk_length_ms=600000):  # 10분 = 600000 밀리초
    audio = AudioSegment.from_file(audio_file)
    chunks = [audio[i:i + chunk_length_ms] for i in range(0, len(audio), chunk_length_ms)]
    output_files = []
    for idx, chunk in enumerate(chunks):
        chunk_name = f"chunk{idx}.wav"
        chunk.export(chunk_name, format="wav")
        output_files.append(chunk_name)
    return output_files

chunks = split_audio(audio_file)
```

```python
from spleeter.separator import Separator

separator = Separator('spleeter:2stems')

def separate_chunks(chunk_files, output_dir='output'):
    for chunk_file in chunk_files:
        separator.separate_to_file(chunk_file, output_dir)
        os.remove(chunk_file)  # 분리 후 원본 chunk 파일 삭제

separate_chunks(chunks)
```

```python
def combine_audio_files(output_dir='output', output_file='combined_output.wav'):
    combined = AudioSegment.empty()
    for root, dirs, files in os.walk(output_dir):
        print(root,files)
        # print(sorted(files))
        for file in sorted(files):
            if file.endswith('accompaniment.wav'):  # 해설자 목소리만 합치기
                audio_segment = AudioSegment.from_file(os.path.join(root, file))
                combined += audio_segment
    combined.export(output_file, format='wav')

combine_audio_files()
```
